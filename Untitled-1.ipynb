```python
from autogen import UserProxyAgent,ConversableAgent,AssistantAgent


import fitz  # PyMuPDF
import os
import uuid

def extract_text_and_images_from_pdf(pdf_path: str, output_dir: str ="pdf_extracted"):
    """
    Extracts text and images from a PDF file.

    Args:
        pdf_path (str): Path to the PDF file.
        output_dir (str): Directory to save extracted images.

    Returns:
        dict: {
            "text": full_text,
            "images": [list of saved image file paths]
        }
    """

    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Open the PDF
    doc = fitz.open(pdf_path)
    all_text = ""
    image_paths = []

    for page_num, page in enumerate(doc):
        # Extract text
        all_text += f"\n\n--- Page {page_num + 1} ---\n"
        all_text += page.get_text()

        # Extract images
        for img_index, img in enumerate(page.get_images(full=True)):
            xref = img[0]
            base_image = doc.extract_image(xref)
            image_bytes = base_image["image"]
            image_ext = base_image["ext"]
            image_name = f"{uuid.uuid4()}.{image_ext}"
            image_path = os.path.join(output_dir, image_name)

            # Save image
            with open(image_path, "wb") as img_file:
                img_file.write(image_bytes)

            image_paths.append(image_path)

    return {
        "text": all_text,
        "images": image_paths
    }


config_list  = [
    {
        'model':"gpt-4o",
        'api_key': ""
    }
]


function_list = [
  {
    "name": "extract_text_and_images_from_pdf",
    "description": "Exract text and images from PDF document",
    "parameters": {
      "type": "object",
      "properties": {
        "title": {
          "type": "string",
          "description": "PDF document for extracting data from",
        }
      },
      "required": ["title"],
    },
  }
]
# Configure Large Language Model (LLM)
llm_config = {
        "seed": 44,  # Set seed for reproducibility
        "config_list": config_list,  # Pass configuration list to LLM
        "temperature": 0  # Set temperature for model generation
    }


humanresearch = UserProxyAgent(
    name="Humanresearcher",
    # system_message=
    llm_config=False,
    code_execution_config=False,
    human_input_mode="NEVER",
)

readeragent = AssistantAgent(
    name="ReaderAgent",
    system_message='''You are a research assistant specialized in reading academic papers and summarizing 
    them concisely. Focus on identifying the key contributions, methodologies, results, and conclusions. 
    Ensure your summary is accurate, structured, and written in a professional and objective tone.
    Avoid personal opinions.''',
    llm_config= {
        "seed": 44,
        "tools": [ 
        {
            "type": "function",
            "function": function_list[0],  # Reuse your existing function definition
        }
    ],
        # "functions": function_list,
        "config_list": config_list,
        "timeout": 280,
        "temperature": 0,
    },  
    code_execution_config=False,
    human_input_mode="NEVER"
)


from autogen.agentchat import register_function

register_function(
  extract_text_and_images_from_pdf,
  caller=readeragent,
  executor=humanresearch,
  name="extract_text_and_images_from_pdf",
  description="Exract text and images from PDF document",
)

chat_result = humanresearch.initiate_chat(
  readeragent,
  message="Get the content of the PDF document from  '/home/saireddy/Downloads/autogapp/1506.02640v5.pdf'. Then, summarize the page content.",
)

print(chat_result)

# tool_agent = ConversableAgent(
#     name="ToolAgent",
#     system_message='''You are a tool-enabled assistant responsible for extracting structured content 
#                       from PDF files. Your job is to accurately retrieve and return raw or cleaned-up text, 
#                       tables, figures, captions, and other relevant metadata. Maintain high fidelity to the 
#                       original formatting. Be efficient, precise, and avoid interpretations.''',
#     llm_config=llm_config,
#     code_execution_config=False,
#     human_input_mode="NEVER",
#     # is_termination_msg=lambda msg: "I gotta go" in msg["content"],
# )
```
